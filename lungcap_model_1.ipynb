```python
import pandas as pd
```

```python
lcn = pd.read_csv(r"C:\Users\HP\Downloads\LungCapData.csv")
```

```python
lcn.head(2)
```

```python
lcn.isnull().sum()
```

```python
lcn['Smoke'] = lcn['Smoke'].replace({'no':0 , 'yes':1})
```

```python
lcn['Gender'] = lcn['Gender'].replace({'female':0, 'male':1})
```

```python
lcn['Caesarean'] = lcn['Caesarean'].replace({'no':0, 'yes':1})
```

```python
lcn.head()
```

```python
# to build a predictive model which will predict the lung capacity
```

```python
from sklearn.model_selection import train_test_split
lcn_train , lcn_test = train_test_split(lcn, test_size= .2)
lcn_train_x = lcn_train.iloc[: , 1::]
lcn_train_y = lcn_train.iloc[: , 0]

lcn_test_x = lcn_test.iloc[: , 1::]
lcn_test_y = lcn_test.iloc[: , 0]
```

```python
from sklearn.linear_model import LinearRegression
linreg = LinearRegression()
linreg.fit(lcn_train_x , lcn_train_y)
Rsqu= linreg.score(lcn_train_x , lcn_train_y)
Rsqu
```

```python
N = lcn_train_x.shape[0]
K = lcn_train_x.shape[1]
AdjRsquare = 1 -(1-Rsqu) *(N-1) / (N-K-1)
AdjRsquare
```
